# Range

```python
for temp in range(5):
    print(f"The number is: {temp}")
```

```python
for temp in range(1,11):
    print(f"The number is: {temp}")
```

```python
for temp in range(1,15,2): #start,stop,step
    print(f"The number is: {temp}")
```

# Nested Loop

```python
First_Name = ['Atta','Ahmad','Waqas','Haider']
Last_Name = ['Ur Rehman','Raza','Ahmad','Maqsood']
for f_name in First_Name:
    for l_name in Last_Name:
        print(f"{f_name} {l_name}")

    print("****************")
```

```python
list1 =[
    [1,2,4,5],
    [9,5,3,6]
]
list2 =[
    [5,6,7,8],
    [2,4,6,8]
]
for i in range(len(list1)):
    for j in range (len(list1[i])):
        # print (f"{list1[i][j]} --> {list2[i][j]}")
        print(f"{list1[i][j]} + {list2[i][j]} = {list1[i][j]+list2[i][j]}")
```

# Dictionary

- dictionary is noting but key-value pair

```python
Bio_data_lst = ['Atta','4.6','20','HSM Colony']
Bio_data_dic  = {
    'Name': 'Atta',
    'CGPA': '4.6',
    'Age': '20',
    'Address': 'HSM Colony'
}
```

```python
Bio_data_dic
```

### Access Values

- Use the key to get value:

```python
Bio_data_lst[0]  # extract value in list
Bio_data_dic['Name']  # extract value in dictionary
```

# Methods of Dictionary

### Add New Data

```python
Bio_data_lst.append('0313-1234567')  # add value in list
Bio_data_dic['Phone'] = '0313-1234567'  # add value in dictionary
```

```python
Bio_data_dic
```

### Change Value

```python
Bio_data_lst = '3.00' # update value in list
Bio_data_dic ['CGPA']= '3.00' # Update value in dictinary
```

```python
Bio_data_dic
```

### Remove Data

```python
Bio_data_lst= ('0313-1234567') # remove value in list
Bio_data_dic.pop('Phone') # remove value in dictionary
```

```python
Bio_data_dic
```

```python
Bio_data_dic.keys() # to get all keys in dictionary
```

```python
Bio_data_dic.values() # to get all values in dictionary
```

```python
Bio_data_dic.items() # to get all items in dictionary
```

###  Apply for loop on dictionary bio_data_dic


```python
for key in Bio_data_dic.keys():
    print(key)
```

```python
for value in Bio_data_dic.values():
    print(value)
```

```python
for data in Bio_data_dic.items():
    print(data)
```

```python
for key, value in Bio_data_dic.items():
    print(f"{key} --> {value}")
```
