# File Handling

```
with open(<file_name>, <mode>) as <alias>:
    <alias>.write("")
```

````python
with open("greeting.txt", "w") as File:
    File.write("Assalam-o-Alaikum")
````

````python
with open("greeting.txt", "r") as File:
    content = File.read()
print(content)
````

````python
with open("greeting.txt", "a") as File:
    File.write("\n Wa-Alaikum-Salam")
````

# CSV Files

- read(r) -> used for reading a file
- write(w) -> used for writing to a file (overwrites existing content)
- append(a) -> used for appending to a file

````python
import csv
with open("competitions.csv") as csvfile:
    reader = csv.reader(csvfile)
    print(list(reader))
````

````python
import csv
with open("competitions.csv") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
````

````python
import csv
with open("competitions.csv") as file:
    reader = csv.reader(file)
    empty_list= []
    for row in reader:
        empty_list.append(row)
print(empty_list)
````

# Exception Handling

````python
num = input("Enter a number:")
print(f"Multiplication of table {num} is:")
try:
    for i in range(1,11): 
        print(f"{int(num)} X {i} = {int(num)*i}")
except Exception as e:
    print(e)
print("Some important code ")
print("End of program")
````

````python
num = input("Enter a number:")
print(f"Multiplication of table {num} is:")
try:
    for i in range(1,11): 
        print(f"{int(num)} X {i} = {int(num)*i}")
except:
    print("Invalid input!")
print("Some important code ")
print("End of program")
````

````python
try:
    num = int(input("Enter a number: "))
    a= [1,2]
    print(a[num])
except ValueError:
    print("Number entered is not integer!")
except IndexError:
    print("index error!")
````
