Q1 — Basic Dictionary Creation

```python
student = {
    "name": "Atta",
    "age": 20,
    "course": "BSCS",
    "city": "Lahore"
}


print("Name:", student["name"])
print("Age:", student["age"])
print("Course:", student["course"])
print("City:", student["city"])
```

Q2 — Adding & Updating Values

```python
car = {
    "brand": "Toyota",
    "model": "Corolla",
    "year": 2018
}


car["color"] = "White"


car["year"] = 2022

print(car)
```

Q3 — Loop Through Dictionary

```python
person = {
    "name": "Ali",
    "age": 25,
    "city": "Karachi"
}


for key, value in person.items():
    print(key, ":", value)
```

Q4 — Count Frequency of Numbers

```python
numbers = [1,2,2,3,3,3,4,4,4,4]

frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

print(frequency)
```

Q5 — Word Counter

```python
sentence = input("Enter a sentence: ")

words = sentence.split()
word_count = {}

for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1

print(word_count)
```

Q6 — Student Marks & Average

```python
marks = {"Math":80, "English":70, "Physics":90}

total = sum(marks.values())
average = total / len(marks)

print("Total Marks:", total)
print("Average Marks:", average)
```

Q7 — Highest Scoring Student

```python
students = {
    "Ali":85,
    "Sara":92,
    "Ahmed":78,
    "Zara":95
}

highest_student = max(students, key=students.get)
highest_marks = students[highest_student]

print("Highest Scoring Student:", highest_student)
print("Marks:", highest_marks)
```

Q8 — Merge Dictionarie

```python
dict1 = {"a":1,"b":2}
dict2 = {"c":3,"d":4}


merged = dict1.copy()
merged.update(dict2)

print(merged)
```

Q9 — Mini Project: Contact Book

```python
contact_book = {}


for i in range(3):
    name = input("Enter name: ")
    phone = input("Enter phone number: ")
    contact_book[name] = phone


search_name = input("Enter name to search: ")

if search_name in contact_book:
    print("Phone Number:", contact_book[search_name])
else:
    print("Contact not found")
```
