# Set

#### Creating set

```python
my_set = {1,2,3,4,5,5,7,8,5,4}
print(my_set)
```

```python
my_set= set([1,2,3,4,5]) # set create using set() function
```

### Add an Element

```python
my_set.add(6)
```

```python
print(my_set)
```

### Remove an Element

```python
my_set.remove(8)
```

```python
print(my_set)
```

### Union (Combine Sets)

```python
a= {1,3,5,7,9}
b= {3,7,2,8,1}

print(a.union(b)) # union of two sets
```

#### Intersection (Common Elements)

```python
a= {1,3,5,7,9}
b= {3,7,2,8,1}

print(a.intersection(b)) # intersection of two set
```

```python
print(a.difference(b)) # difference of two set
```

# Match Statement

```python
Day = 5
match Day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
```

```python
day = 2
match day:
    case 1|7:
        print("Weekend")
    case _:
        print("Weekday")
```

```python
num = 10
match num:
    case 1:
        print("one")
    case 2:
        print("two")
    case 3:
        print("three")
    case _:
        print("Invalid number")
```

```python
num = 5
match num:
    case 0:
        print("Zero")
    case 1:
        print("One")
    case _:
        print("other")
```

```python
a= float(input("Enter a number:"))
b = float(input("Enter another number:"))
op= input("Enter an operatoe (+,-,*,/,%)")

match op:
    case "+":
        print("Result:",a+b)
    case "-":
        print("Result:",a-b)
    case "*":
        print("Result:",a*b)
    case "/":
        print("Result:",a/b)
    case "%":
        print("Result:",a%b)
```
