## Positional Argumemts

````python
def bmi_calculaor(name, weight=20,height= 3):
    return (f"{name} BMI is: {weight/(height**2)}")
````

````python
bmi_calculaor('Atta') # here this function will take default arguments
````

````python
bmi_calculaor("Arham", 50, 5.5) 
````

#### Q: Create function that print atleat 2 details on customer as paramter and 3 details as **kwargs

````python
def coustmer(name,age, **kwargs):
    print("Name:",name)
    print("Age:",age)
    print("Other details:")
    for key,value in kwargs.items():
        print(f"{key}: {value}")
````

````python
coustmer('Ali', 21, Email='ali@gmail.com', profession= 'Student')
````
