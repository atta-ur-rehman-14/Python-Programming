### Sum of two Number

````python
def addition(num1,num2):
    sum = num1 + num2
    print(f"The sum of {num1} and {num2} is {sum}")
````

````python
addition(8,5) # Here we are calling the function and passing the values as argument
````

## Positional Argumemts

````python
def bmi_calculaor(name, weight=20,height= 3):
    return (f"{name} BMI is: {weight/(height**2)}")
````

````python
bmi_calculaor('Atta') # here this function will take default arguments
````

````python
bmi_calculaor("Arham", 50, 5.5) 
````

````python
bmi_cal('Raza', 'Ahmad') # its wrong bcz name is showing reverse order
````
