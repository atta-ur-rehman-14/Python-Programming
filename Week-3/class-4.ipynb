# conditional statements
- if internet avaliable --> Class will be held
- if internet not avaliable --> Class will not be conducted

Syntax:
if<condition>:
    <body of if>
else:
    <body of else>

```python
internet_avaliable = False
independence_day = "no"

if internet_avaliable and independence_day == "yes":
    print("watch independence event:")
elif internet_avaliable:
    print("class will be held online")
else:
    print("class will not held")
```

```python
marks = 40

if marks > 80:

    print("Your Grade is A")

elif marks > 60 and marks <= 80:

    print("Your Grade is B")

elif marks > 50 and marks <= 60:

    print("Your Grade is C")
elif marks > 40 and marks <=30:
    print("Your Grade is D")
elif marks > 30 and marks <=40:
    print("Your grade is E")
```
