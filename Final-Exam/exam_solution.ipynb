# C1  Function Basics & Return Values
Write the following standalone utility functions, each with a docstring:
- is_prime(n) — returns True if n is prime, False otherwise
- celsius_to_all(c) — returns a tuple of (fahrenheit, kelvin)
- count_vowels(text) — returns the count of vowels in a string
- power(base, exp=2) — returns base raised to exp. Default exponent is 2.

Call each function with at least one test case and print the result.


```python
def is_prime(n):
    """Returns True if n is prime, False otherwise."""
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

def celsius_to_all(c):
    """Returns a tuple of (fahrenheit, kelvin)."""
    fahrenheit = (c * 9/5) + 32
    kelvin = c + 273.15
    return (fahrenheit, kelvin)

def count_vowels(text):
    """Returns the count of vowels in a string."""
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char in vowels:
            count += 1
    return count

def power(base, exp=2):
    """Returns base raised to exp. Default exponent is 2."""
    return base ** exp


print("is_prime(11): {}".format(is_prime(11)))

temps = celsius_to_all(25)
print("25°C in Fahrenheit and Kelvin: {}".format(temps))

vowel_count = count_vowels("Hello World")
print("Vowels in 'Hello World': {}".format(vowel_count))


print("power(5): {}".format(power(5)))        # Uses default exp=2
print("power(2, 3): {}".format(power(2, 3)))  # Uses exp=3
```

# C4  Recursion
Part 1 — Fibonacci (2 marks):
Write fibonacci(n) recursively. Print the first 10 Fibonacci numbers.

Part 2 — Factorial (2 marks):
Write factorial(n) recursively. Handle n=0 and negative input gracefully.

Part 3 — Flatten a nested list (2 marks):
Write flatten(lst) that takes a deeply nested list and returns a flat list using recursion.
flatten([1, [2, [3, 4], 5], [6, 7]])
##### Expected: [1, 2, 3, 4, 5, 6, 7]


```python
def fibonacci(n):
    """Returns the nth Fibonacci number recursively."""
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

print("First 10 Fibonacci numbers:")
for i in range(10):
    print(fibonacci(i), end=" ")
print("\n")


def factorial(n):
    """Returns factorial of n recursively. Handles 0 and negative inputs."""
    if n < 0:
        return "Error: Negative input"
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print("Factorial of 5: {}".format(factorial(5)))
print("Factorial of 0: {}".format(factorial(0)))
print("Factorial of -1: {}".format(factorial(-1)))
print()


def flatten(lst):
    """Takes a deeply nested list and returns a flat list using recursion."""
    flat_list = []
    for item in lst:
        if type(item) == list:
        
            flat_list.extend(flatten(item))
        else:
           
            flat_list.append(item)
    return flat_list

nested = [1, [2, [3, 4], 5], [6, 7]]
print("Original nested list: {}".format(nested))
print("Flattened list: {}".format(flatten(nested)))
```

  - available_books() — returns all currently available books
  - total_books() — returns count of all books

Create at least 4 Book objects, add to Library, checkout 2, then display all available books.



```python
class Book:
    def __init__(self, title, author, isbn, is_available=True):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_available = is_available

    def checkout(self):
        if self.is_available:
            self.is_available = False
            return True
        return False

    def return_book(self):
        self.is_available = True

    def __str__(self):
        status = "Available" if self.is_available else "Checked Out"
        return "{} by {} (ISBN: {}) - [{}]".format(self.title, self.author, self.isbn, status)


class Library:
    def __init__(self):
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def find_by_author(self, author):
        return [book for book in self.books if book.author.lower() == author.lower()]

    def available_books(self):
        return [book for book in self.books if book.is_available]

    def total_books(self):
        return len(self.books)



my_library = Library()


b1 = Book("Python Crash Course", "Eric Matthes", "978-1593279288")
b2 = Book("Automate the Boring Stuff", "Al Sweigart", "978-1593279929")
b3 = Book("Data Science from Scratch", "Joel Grus", "978-1492041139")
b4 = Book("Deep Learning with Python", "Francois Chollet", "978-1617294433")

my_library.add_book(b1)
my_library.add_book(b2)
my_library.add_book(b3)
my_library.add_book(b4)


b1.checkout()
b3.checkout()


print("Total books in library: {}".format(my_library.total_books()))

print("\nCurrently Available Books:")
available = my_library.available_books()
for book in available:
    print(book)


print("\nSearching for books by 'Al Sweigart':")
search_results = my_library.find_by_author("Al Sweigart")
for book in search_results:
    print(book)
```

# D4  File Handling
Write a script that manages a text-based student record file called students.txt.
   - save_student(name, marks) — appends a student record in format: Ali,85
   - load_students() — reads file and returns list of dicts: [{"name": "Ali", "marks": 85}, ...]
   - top_student() — reads file and returns name of student with highest marks

Save at least 5 students, then call top_student() and print the result.


```python
def save_student(name, marks):
    """Appends a student record to students.txt in format: Name,Marks"""
    with open("students.txt", "a") as f:

        f.write("{},{}\n".format(name, marks))

def load_students():
    """Reads file and returns a list of dictionaries."""
    student_list = []
    try:
        with open("students.txt", "r") as f:
            for line in f:
            
                parts = line.strip().split(",")
                if len(parts) == 2:
                    name = parts[0]
                    marks = int(parts[1])
                    student_list.append({"name": name, "marks": marks})
    except FileNotFoundError:
        print("Error: The file 'students.txt' was not found.")
    return student_list

def top_student():
    """Reads file and returns name of student with highest marks."""
    students = load_students()
    if not students:
        return "No records found"
    
    best_student = max(students, key=lambda x: x["marks"])
    return best_student["name"]


save_student("Ali", 85)
save_student("Sara", 92)
save_student("Umar", 78)
save_student("Hina", 95)
save_student("Bilal", 88)


winner = top_student()
print("The top student is: {}".format(winner))

print("\nFull Records List:")
print(load_students())
```

```python
import os
class Student:  # Feature 1: Student Class 
    def __init__(self, roll_no, name, marks_dict):
        self.roll_no = roll_no
        self.name = name
        self.marks = marks_dict  # Dictionary: {"Math": 85, "Physics": 78}

    def average(self):
        """Returns the average of all subject marks."""
        if not self.marks:
            return 0
        return sum(self.marks.values()) / len(self.marks)

    def grade(self):
        """Returns A/B/C/D/F based on average."""
        avg = self.average()
        if avg >= 90: return "A"
        elif avg >= 80: return "B"
        elif avg >= 70: return "C"
        elif avg >= 60: return "D"
        else: return "F"

    def __str__(self):
        """Returns a neat single-line summary."""
        return "Roll: {} | Name: {} | Avg: {:.2f} | Grade: {}".format(
            self.roll_no, self.name, self.average(), self.grade()
        )

FILENAME = "records.txt" # Feature 2: Data Storage & File Persistence

def save_all(students):
    """Writes all student records to file in CSV-like format."""
    with open(FILENAME, "w") as f:
        for s in students:
        
            marks_str = ",".join(["{}:{}".format(k, v) for k, v in s.marks.items()]) # Format: Roll,Name,Sub1:Score,Sub2:Score
            f.write("{},{},{}\n".format(s.roll_no, s.name, marks_str))

def load_all():
    """Reads file and returns list of Student objects."""
    students = []
    if not os.path.exists(FILENAME):
        return students
    
    try:
        with open(FILENAME, "r") as f:
            for line in f:
                parts = line.strip().split(",")
                if len(parts) >= 3:
                    roll = parts[0]
                    name = parts[1]
                    
                    marks = {} # Reconstruct marks dictionary
                    for pair in parts[2:]:
                        sub, score = pair.split(":")
                        marks[sub] = float(score)
                    students.append(Student(roll, name, marks))
    except Exception as e:
        print("Error loading data: {}".format(e))
    return students


def get_valid_marks():  # Feature 3: Menu System & Input Validation 
    subjects = ["Math", "Physics", "Python"]
    marks = {}
    for sub in subjects:
        while True:
            val = input("Enter marks for {} (0-100): ".format(sub))
            if val.replace('.', '', 1).isdigit():
                score = float(val)
                if 0 <= score <= 100:
                    marks[sub] = score
                    break
            print("Invalid input. Enter a number between 0 and 100.")
    return marks

def main_menu():
    students = load_all()
    
    while True:
        print("\n=== Student Management System ===")
        print("1. Add new student")
        print("2. View all students")
        print("3. Search by roll number")
        print("4. Delete a student")
        print("5. Show top performer")
        print("6. Show class statistics")
        print("0. Exit")
        
        choice = input("Select an option: ")
        
        if choice == "1":
            roll = input("Enter Roll Number: ").strip()

            if any(s.roll_no == roll for s in students):
                print("Error: Roll number already exists!")
                continue
            name = input("Enter Name: ").strip()
            if not name:
                print("Error: Name cannot be empty!")
                continue
            marks = get_valid_marks()
            students.append(Student(roll, name, marks))
            save_all(students)
            print("Student added successfully!")

        elif choice == "2":
            if not students: print("No records found.")
            for s in students: print(s)

        elif choice == "3":
            roll = input("Enter Roll Number to search: ")
            found = False
            for s in students:
                if s.roll_no == roll:
                    print("Found: {}".format(s))
                    found = True
                    break
            if not found: print("Student not found.")

        elif choice == "4":
            roll = input("Enter Roll Number to delete: ")
            initial_len = len(students)
            students = [s for s in students if s.roll_no != roll]
            if len(students) < initial_len:
                save_all(students)
                print("Student deleted.")
            else:
                print("Roll number not found.")

        elif choice == "5":
            if students:
                top = max(students, key=lambda s: s.average())
                print("Top Performer: {}".format(top))
            else:
                print("No data available.")

        elif choice == "6":
            
            if not students: # Feature 4: Analytics
                print("No data available for statistics.")
                continue
            
            total_students = len(students)
            class_avg = sum(s.average() for s in students) / total_students
            highest = max(students, key=lambda s: s.average())
            lowest = min(students, key=lambda s: s.average())
            
            grades = {"A": 0, "B": 0, "C": 0, "D": 0, "F": 0}
            subject_totals = {}
            
            for s in students:
                grades[s.grade()] += 1
                for sub, score in s.marks.items():
                    subject_totals[sub] = subject_totals.get(sub, 0) + score

            print("\n--- Class Statistics ---")
            print("Total Students: {}".format(total_students))
            print("Class Average: {:.2f}".format(class_avg))
            print("Highest Scorer: {} ({:.2f})".format(highest.name, highest.average()))
            print("Lowest Scorer: {} ({:.2f})".format(lowest.name, lowest.average()))
            print("Grade Distribution: {}".format(grades))
            
            
            best_sub = max(subject_totals, key=lambda k: subject_totals[k]) # Highest average subject
            print("Top Subject (Highest Total): {}".format(best_sub))

        elif choice == "0":
            save_all(students)
            print("Data saved. Goodbye!")
            break
        else:
            print("Invalid choice. Try again.")

if __name__ == "__main__":
    main_menu()
```
