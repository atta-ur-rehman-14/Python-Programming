# Loops

````python
loop = [1,2,3,4,5] 
````

### Syntax

```bash
for <temp-variable> in <iterator>:
    <for loop body>
```

````python
for temp in loop:
    print(temp)
````

````python
for temp in loop[::-1]:
    print(temp)
````

````python
for i in loop[1::2]:
    print(i)
````

````python
for i in loop[::2]:
    print(i)
````

### Print Table of 5

````python

for i in range (1,11):
    print("5 x " , i , " = " , 5*i)
````

# While Loop

```bash
while <condition>:
    <while body>
```

````python
num = int(input("Enter the number:"))
while (num<=10):
    num = int(input("Enter the number:"))
    print(num)
````

### Range..

````python
for temp in range(5):
    print(f"The number is: {temp}")
````

````python
for temp in range(1,11):
    print(f"The number is: {temp}")
````

````python
for temp in range(1,15,2): #start,stop,step
    print(f"The number is: {temp}")
````

# Nested Loop

````python
First_Name = ['Atta','Ahmad','Waqas','Haider']
Last_Name = ['Ur Rehman','Raza','Ahmad','Maqsood']
for f_name in First_Name:
    for l_name in Last_Name:
        print(f"{f_name} {l_name}")

    print("****************")
````
