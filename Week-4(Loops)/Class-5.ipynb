# Loops

````python
loop = [1,2,3,4,5] 
````

### Syntax

```bash
for <temp-variable> in <iterator>:
    <for loop body>
```

````python
for temp in loop:
    print(temp)
````

````python
for temp in loop[::-1]:
    print(temp)
````

````python
for i in loop[1::2]:
    print(i)
````

````python
for i in loop[::2]:
    print(i)
````

### Print Table of 5

````python

for i in range (1,11):
    print("5 x " , i , " = " , 5*i)
````
