#### Task: pirnt all details of student 1 and student 2

```python
students = {
    "student1": {
        "name": "Alice",
        "age": 20,
        "grades": {
            "math": 90,
            "science": 85,
            "english": 88
        }
    },
    "student2": {
        "name": "Bob",
        "age": 22,
        "grades": {
            "math": 75,
            "science": 80,
            "english": 78
        }
    }
}
```

```python
for i in students:
    for key,value in students[i].items():
        print(f"{key} --> {value}")
```
