# Classes Topic Car Eample
- Characteristics/ Properties (Constructor: should be only one per class)

    - Type
    - Model
    - Make (years)
    - color
- Behaviours/ Actions (Methods: can be more than one)

    - can travel
    - details

```python
class car():
    def __init__(self,type,model,year,color): # constructor
        self.type = type # object variables
        self.model = model
        self.year = year 
        self.color = color
    def can_travel(self): # method 1 / function
        print("Automobiles car can help us to travel")
    def car_details(self):  # method 2
        print(f"--- Car Characterisics ----")
        print(f"Type is:{self.type}")
        print(f"Model is:{self.model}")
        print(f"Year is:{self.year}")
        print(f"Color is:{self.color}")



car_obj = car("SUV","Honda",2023,"Black")
car_obj.can_travel()
```

```python
## characteristics
print(car_obj.type)
print(car_obj.model)
print(car_obj.year)
print(car_obj.color)

## methods
print('------- METHODS -----------')

car_obj.can_travel()

car_obj.car_details()
```

# inheritence

- Parent class/base class = in which class is inhereted
- Child class/ derived class=  this inherit parent class

```python
class Employee:
    def __init__(self,name,id):
        self.name = name
        self.id = id
    def showdetails(self):
        print(f"Employee Name is {self.name} and the ID is {self.id}.")
class programmer(Employee): # inheritance
    def showlanguage(self):
        print("The python is programming language.")
Emp_1 = programmer('Atta', 27) # object of prgrammer class that is chlid class of employee class
Emp_1.showdetails()
Emp_1.showlanguage()
```

### Single class Inheritance

- most common form of inheritance is single inheritance

```python
class Animal:
    def __init__(self,name,species):
        self.name = name
        self.species = species
    def make_sound(self):
        print("Animal can make sound")
class Dog(Animal):
    def __init__(self,name,breed):
        Animal.__init__(self,name,species='Dog') # call parent class constructor
        self.breed = breed
    def make_sound(self):
        print("Bark")
D= Dog('Dog','Bulldog')
D.make_sound()
a= Animal('Animal','Mammal')
a.make_sound()
```

# Multiple inheritance
A child class that hai multiple parent class

```python
class Employee:
    def __init__(self,name):
        self.name = name
    def show(self):
        print(f"The name is {self.name}")

class Dancer:
    def __init__(self,dance):
        self.dance = dance
    def show(self):
        print(f"The dance is {self.dance}")

class employee_dancer(Dancer,Employee): # multiple inhertance
    def __init__(self,name,dance):
        self.name = name
        self.dance = dance

emp_dance = employee_dancer("XYZ","Hip Hop")
emp_dance.show()
```

# Multilevel Inhertance

```python
class Animal:
    def eat(self):
        print("Animal eats")

class Dog(Animal):
    def bark(self):
        print("Dog barks")

class Puppy(Dog):
    def weep(self):
        print("Puppy weeps")

p = Puppy()
p.eat()
p.bark()
p.weep()
```

# Abstraction 
- Abstraction is the concept of hiding implementation details and showing only the  essential features of an object to the user.

## for making abstract class
- inherit to ABC
- used at least 1 abstract method in the class
- we cannot create object of Abstract Class

```python
from abc import ABC,abstractmethod
class Bankapp(ABC):
    def database(self):
        print("Connected to Database")
    @abstractmethod
    def security(self):
        pass
```

```python
class MobileApp(Bankapp):
    def mobile_login(self):
        print("Login to Mobile")
    def security(self):
        print("Security for app")
    
```

```python
mob = MobileApp()
```

```python
mob.database()
```

# Encapsulation


### Encapsulation provides several benefits:

- `Data Protection`: Prevents accidental modification of data
- `Validation`: You can validate data before setting it
- `Flexibility`: Internal implementation can change without affecting external code
- `Control`: You have full control over how data is accessed and modified

```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.__age = age

  def get_age(self):
    return self.__age

p1 = Person("Tobias", 25)
print(p1.get_age())
print(p1.get_)
```

### Getter Method
A getter method is used to retrieve (read) the value of a private variable.
### Setter Method
A setter method is used to update (modify) the value of a private variable.

```python
class person:
    def __init__(self,name,age):
        self.name = name
        self.__age= age
    def get_age(self):  # getter Method
        return self.__age
    def set_age(self,age): # setter method
        if age >= 18:
            self.__age = age
        else:
            print("Age must be positive.")
p1 = Person("Ali", 25) 
print(p1.name)     # Access public variable
print(p1.get_age()) # Access private variable

   
```

```python
p1.set_age(18)
print(p1.get_age())
```
